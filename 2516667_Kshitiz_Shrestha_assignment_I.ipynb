## About the Dataset

The Human Development Index (HDI) measures a country’s achievements in key dimensions of human development. The dataset spans the years 1990 to 2022 and includes all countries for which HDI is reported. The main dimensions included in the dataset are:

Health: Life expectancy at birth

Education: Mean years of adult schooling and expected years of schooling for children

Standard of Living: Gross national income (GNI) per capita

These indicators are used to assess differences in development levels across countries and to analyze trends in human development over time.

3. Outlier Detection:

```python
# prepare south aisa dataset

sa_od = south_asia_df[["country", "year", "hdi", "gross_inc_percap"]].dropna()
```

```python
# Compute IQR bounds (South Asia only)
# HDI bounds
Q1_hdi = sa_od["hdi"].quantile(0.25)
Q3_hdi = sa_od["hdi"].quantile(0.75)
IQR_hdi = Q3_hdi - Q1_hdi

lower_hdi = Q1_hdi - 1.5 * IQR_hdi
upper_hdi = Q3_hdi + 1.5 * IQR_hdi

# GNI bounds
Q1_gni = sa_od["gross_inc_percap"].quantile(0.25)
Q3_gni = sa_od["gross_inc_percap"].quantile(0.75)
IQR_gni = Q3_gni - Q1_gni

lower_gni = Q1_gni - 1.5 * IQR_gni
upper_gni = Q3_gni + 1.5 * IQR_gni
```

```python
# flag outliers

sa_od["outlier_hdi"] = (sa_od["hdi"] < lower_hdi) | (sa_od["hdi"] > upper_hdi)
sa_od["outlier_gni"] = (sa_od["gross_inc_percap"] < lower_gni) | (sa_od["gross_inc_percap"] > upper_gni)

sa_od["outlier"] = sa_od["outlier_hdi"] | sa_od["outlier_gni"]

outliers_sa = sa_od[sa_od["outlier"]]
outliers_sa[["country", "year", "hdi", "gross_inc_percap"]]
```

```python
# scatter plot

plt.figure(figsize=(8,6))

# Normal points
plt.scatter(
    sa_od.loc[~sa_od["outlier"], "gross_inc_percap"],
    sa_od.loc[~sa_od["outlier"], "hdi"],
    alpha=0.5,
    label="Normal"
)

# Outliers
plt.scatter(
    sa_od.loc[sa_od["outlier"], "gross_inc_percap"],
    sa_od.loc[sa_od["outlier"], "hdi"],
    alpha=0.9,
    label="Outliers"
)

plt.xlabel("GNI per Capita")
plt.ylabel("HDI")
plt.title("HDI vs GNI per Capita (South Asia) with Outliers")
plt.legend()
plt.tight_layout()
plt.show()
```

5. Gap Analysis:

```python
# checking max value

df['gross_inc_percap'].max()
```

-> gross_inc_percap is in thousands of dollars

-> hdi is between 0 and 1

So subtracting directly gives meaningless huge values.

👉 We should normalize income first so both are comparable.

We’ll use:

GNI_norm = (gross_inc_percap − min) / (max − min) <br>
GNI_HDI_Gap = GNI_norm − hdi


```python
# prepare_data

gap_df = south_asia_df[["country", "year", "hdi", "gross_inc_percap"]].dropna().copy()
```

```python
# normalize GNI

gni_min = gap_df["gross_inc_percap"].min()
gni_max = gap_df["gross_inc_percap"].max()

gap_df["gni_norm"] = (gap_df["gross_inc_percap"] - gni_min) / (gni_max - gni_min)
```

```python
# compute the gap

gap_df["GNI_HDI_Gap"] = gap_df["gni_norm"] - gap_df["hdi"]
gap_df.head()

```

```python
# average gap per country

gap_avg = gap_df.groupby("country")["GNI_HDI_Gap"].mean().sort_values()
gap_avg
```

5. HDI Disparity:

To measure the disparity of HDI within each region, we compute the range and the coefficient of variation (CV).

The range measures the spread between the highest and lowest HDI values in a region. It is calculated as:

Range = max(HDI) − min(HDI)

A larger range indicates a greater difference between the most and least developed countries in the region.

The coefficient of variation (CV) measures the relative dispersion of HDI values around the mean. It is calculated as:

CV = standard deviation of HDI / mean of HDI

```python
# compute range

range_sa = south_asia_df["hdi"].max() - south_asia_df["hdi"].min()
range_me = middle_east_df["hdi"].max() - middle_east_df["hdi"].min()

range_sa, range_me
```

```python
# compute coefficient and variation

cv_sa = south_asia_df["hdi"].std() / south_asia_df["hdi"].mean()
cv_me = middle_east_df["hdi"].std() / middle_east_df["hdi"].mean()

cv_sa, cv_me
```

```python
# display

disparity_df = pd.DataFrame({
    "Range": [range_sa, range_me],
    "CV": [cv_sa, cv_me]
}, index=["South Asia", "Middle East"])

disparity_df
```

The Middle East exhibits greater variation in HDI than South Asia, as indicated by its higher HDI range (0.513 vs 0.321) and higher coefficient of variation (0.179 vs 0.154). This suggests larger disparities in development levels among Middle Eastern countries compared to South Asian countries.

7. Outlier Detection:

```python
# detect outliers

def detect_outliers_iqr(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[column] < lower) | (df[column] > upper)]
    return outliers, lower, upper
```

```python
# Detect for each region & variable:

# South Asia
sa_hdi_out, sa_hdi_low, sa_hdi_up = detect_outliers_iqr(south_asia_df, "hdi")
sa_gni_out, sa_gni_low, sa_gni_up = detect_outliers_iqr(south_asia_df, "gross_inc_percap")

# Middle East
me_hdi_out, me_hdi_low, me_hdi_up = detect_outliers_iqr(middle_east_df, "hdi")
me_gni_out, me_gni_low, me_gni_up = detect_outliers_iqr(middle_east_df, "gross_inc_percap")

len(sa_hdi_out), len(sa_gni_out), len(me_hdi_out), len(me_gni_out)
```
